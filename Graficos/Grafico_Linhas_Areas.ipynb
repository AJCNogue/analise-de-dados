```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dias = [1,2,3,4,5,6,7]
temperatura = [22,24,21,25,28,27,29]
```

```python
plt.plot(dias, temperatura, marker='o', linestyle='-', color='darkorange')
plt.title('Temperatura ao Longo da Semana')
plt.xlabel('Dia')
plt.ylabel('Temperatura(°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.fill_between(dias, temperatura, color='skyblue', alpha=0.4)
plt.plot(dias, temperatura, color='Slateblue', alpha=0.8)
plt.title('Volume Acumulado(Área)')
plt.show()
```
