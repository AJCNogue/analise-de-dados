```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
categorias = ['Produto A', 'Produto B', 'Produto C', 'Produto D']
vendas = [150, 230, 180, 310]
```

```python
plt.figure(figsize=(6,4))
plt.bar(categorias, vendas, color='steelblue')
plt.title('Vendas por Produto')
plt.ylabel('Unidades')
plt.show()
```

```python
plt.barh(categorias, vendas, color='teal')
plt.title('Vendas por Produto (Horizontal)')
plt.xlabel('Unidades')
plt.show()
```
