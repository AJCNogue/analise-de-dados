```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(10)
x = np.random.rand(30) * 100
y = x * 1.5 + np.random.normal(0, 15, 30)
tamanho_bolha = np.random.rand(30) * 500
```

```python
plt.scatter(x,y,color='crimson',alpha=0.7)
plt.title('Grafico de Dispersão')
plt.xlabel('Investimento')
plt.ylabel('Retorno')
plt.show()
```

```python
plt.scatter(x,y, s=tamanho_bolha, alpha=0.5, c=tamanho_bolha, cmap='viridis')
plt.colorbar(label='Volume')
plt.title('Grafico de Bolhas (3 Variáveis)')
plt.show()
```
