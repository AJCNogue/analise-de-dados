```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
fatias = [45, 30, 15, 10]
rotulos = ['Aluguel','Alimentação','Transporte','Lazer']
```

```python
plt.pie(fatias, labels=rotulos, autopct='%1.1f%%', startangle=90)
plt.title('Distribuição de Gastos (Pizza)')
plt.show()
```

```python
plt.pie(fatias, labels=rotulos, autopct='%1.1f%%', startangle=90, wedgeprops=dict(width=0.4))
plt.title('Distribuição de Gastos (Rosca)')
```
