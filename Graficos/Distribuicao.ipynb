```python
import pandas as np
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
dados = np.random.normal(loc=50, scale=10, size=200)
```

```python
sns.histplot(dados, kde=True, bins=15, color='royalblue')
plt.title('Histograma de Distribuição')
plt.show()
```

```python
sns.boxplot(x=dados, color='lightgreen')
plt.title('Boxplot (Mediana, Quartis e Outliers)')
plt.show()
```

```python
sns.violinplot(x=dados, color='coral')
plt.title('Gráfico de Violino')
plt.show()
```
